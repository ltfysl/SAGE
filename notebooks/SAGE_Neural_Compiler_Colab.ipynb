# SAGE Neural Compiler Training

Train a neural SAGE compiler (sentence-level) using T5 on synthetic data.

**Expected runtime on free T4 GPU:** ~30–35 minutes total.

**Steps:**
1. Setup — clone repo, install dependencies
2. Generate synthetic training data
3. Train the sentence compiler
4. Evaluate on benchmarks
5. View results
6. Save trained model to Google Drive (optional)

```python
# Clone and setup
!git clone https://github.com/your-repo/SAGE.git /content/SAGE 2>/dev/null || echo "Using uploaded SAGE"
%cd /content/SAGE
!pip install -q torch transformers datasets accelerate scikit-learn typer rich pydantic pandas
import sys; sys.path.insert(0, 'src')
```

```python
# Generate neural training data
!python -m sage.cli generate-neural-training-data --sentence-samples 30000 --story-samples 10000
!python -m sage.cli generate-data --samples 5000 --max-events 50 --benchmarks
```

```python
# Train neural sentence compiler
!python -m sage.cli train-neural-sentence-compiler \
  --train data/generated/neural_sentence_train.jsonl \
  --dev data/generated/neural_sentence_dev.jsonl \
  --model google/flan-t5-small \
  --epochs 3 --batch-size 8 \
  --max-input-length 256 --max-target-length 256 \
  --output-dir models/neural-sentence-compiler
```

```python
# Evaluate on all benchmarks
!python -m sage.cli eval-neural-sentence-compiler \
  --model models/neural-sentence-compiler \
  --benchmarks all
```
